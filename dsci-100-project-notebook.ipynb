#### Task 5.2 Perform the Analysis (20 points)

- Perform the proposed analysis.
- Be sure to comment your code throughout so that it is easy to follow.

</div>

```python
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.pipeline import make_pipeline
from sklearn.compose import make_column_transformer
```

```python
#Split the tidy dataset into 75% training data and 25% testing data
basketball_train, basketball_test = train_test_split(
    basketball_tidy,
    test_size=0.25,
    stratify=basketball_tidy['bid']
)
basketball_train.head(6)
    
```

```python
#Seperate predictors and target variable in the training data and testing data
X_train = basketball_train[['conf_wins_pct', 'conf_rank']]
y_train = basketball_train['bid']

X_test = basketball_test[['conf_wins_pct', 'conf_rank']]
y_test = basketball_test['bid']
```

```python
#standardize the predictor variables
basketball_preprocessor = make_column_transformer(
    (StandardScaler(), ['conf_wins_pct', 'conf_rank'])
)
#create a KNN classification pipeline
basketball_pipe = make_pipeline(
    basketball_preprocessor,
    KNeighborsClassifier()
)
parameter_grid = {
    'kneighborsclassifier__n_neighbors': range(1, 301, 5)
}
#tune the classifier
basketball_grid = GridSearchCV(
    estimator=basketball_pipe,
    param_grid=parameter_grid,
    cv=5
)
#fit the model
basketball_grid.fit(X_train, y_train)
accuracies_grid = pd.DataFrame(basketball_grid.cv_results_)
accuracies_grid
```

```python
#store the cross-validation results
accuracies_grid = (
    accuracies_grid[[
        'param_kneighborsclassifier__n_neighbors',
        'mean_test_score'
    ]].rename(columns={
        'param_kneighborsclassifier__n_neighbors':'n_neighbors'
    })
)
accuracies_grid
```

```python
#find the best number of neighbors
basketball_grid.best_params_
```

```python
#plot line graph
accuracy_vs_k = alt.Chart(accuracies_grid).mark_line(point=True).encode(
    x=alt.X('n_neighbors').title('Number of Neighbors(K)'),
    y=alt.Y('mean_test_score').scale(zero=False).title('Cross-validation Accuracy')
)
accuracy_vs_k
```

```python
#predict the class labels in the test set
basketball_test['predicted'] = basketball_grid.predict(X_test)
basketball_test
```

```python
#calculate the test accuracy
basketball_grid.score(X_test, y_test)
```
